```python
import numpy as np

x = np.array([6, 20])

w = np.array([5, 0.2])

b = 10

z = np.dot(w, x) + b

print("Weighted sum:", z)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

output = sigmoid(z)

print("Neuron output:", output)
```

```python
def relu(z):
    return np.maximum(0, z)

for value in [-3, -1, 0, 2, 5]:
    print(value, "→", relu(value))
```
