```python
import numpy as np

X = np.array([
    [1, 2, 20],
    [1, 4, 20],
    [1, 6, 30]
])

y = np.array([50, 70, 90])

beta = np.linalg.inv(X.T @ X) @ X.T @ y

beta
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X[:, 1:], y)

print("Sklearn coefficients:", model.coef_)
print("Sklearn intercept:", model.intercept_)
```

```python
from sklearn.metrics import mean_absolute_error

y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

train_mae = mean_absolute_error(y_train, y_train_pred)
test_mae = mean_absolute_error(y_test, y_test_pred)

print("Training MAE:", train_mae)
print("Test MAE:", test_mae)
```

```python
from sklearn.preprocessing import PolynomialFeatures
X = df[["Study_Hours", "Age"]]

poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)

print("Original shape:", X.shape)
print("Polynomial shape:", X_poly.shape)
print("Feature names:", poly.get_feature_names_out(X.columns))
```

```python
X_poly_train, X_poly_test, y_poly_train, y_poly_test = train_test_split(
    X_poly, y, test_size=0.2, random_state=42
)

poly_model = LinearRegression()
poly_model.fit(X_poly_train, y_poly_train)

y_poly_train_pred = poly_model.predict(X_poly_train)
y_poly_test_pred = poly_model.predict(X_poly_test)

poly_train_mae = mean_absolute_error(y_poly_train, y_poly_train_pred)
poly_test_mae = mean_absolute_error(y_poly_test, y_poly_test_pred)

print("Polynomial Training MAE:", poly_train_mae)
print("Polynomial Test MAE:", poly_test_mae)
```

```python
poly10 = PolynomialFeatures(degree=10)

X_poly10 = poly10.fit_transform(X)

X_train10, X_test10, y_train10, y_test10 = train_test_split(
    X_poly10, y, test_size=0.2, random_state=42
)

model10 = LinearRegression()
model10.fit(X_train10, y_train10)

y_train10_pred = model10.predict(X_train10)
y_test10_pred = model10.predict(X_test10)

train_mae10 = mean_absolute_error(y_train10, y_train10_pred)
test_mae10 = mean_absolute_error(y_test10, y_test10_pred)

print("Degree 10 Training MAE:", train_mae10)
print("Degree 10 Test MAE:", test_mae10)
```

```python
from sklearn.model_selection import cross_val_score

for degree in range(1, 11):
    poly = PolynomialFeatures(degree=degree)
    X_poly_degree = poly.fit_transform(X)

    model = LinearRegression()

    scores = cross_val_score(
        model,
        X_poly_degree,
        y,
        cv=5,
        scoring="neg_mean_absolute_error"
    )

    mean_mae = -scores.mean()

    print(f"Degree {degree}: CV MAE = {mean_mae:.3f}")
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ("poly", PolynomialFeatures()),
    ("model", LinearRegression())
])

param_grid = {
    "poly__degree": range(1, 11)
}

grid_search = GridSearchCV(
    pipeline,
    param_grid,
    cv=5,
    scoring="neg_mean_absolute_error"
)

grid_search.fit(X, y)

print("Best Degree:", grid_search.best_params_["poly__degree"])
print("Best CV MAE:", -grid_search.best_score_)
```

```python
best_model = grid_search.best_estimator_

y_test_pred = best_model.predict(X_test)

final_test_mae = mean_absolute_error(y_test, y_test_pred)

print("Best Degree:", grid_search.best_params_["poly__degree"])
print("CV MAE:", -grid_search.best_score_)
print("Final Test MAE:", final_test_mae)
```

```python
from sklearn.linear_model import LogisticRegression

X_class = df[["Study_Hours"]]

y_class = (df["Marks"] >= 70).astype(int)

print("X:")
print(X_class.head())

print("\ny:")
print(y_class.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

X_train, X_test, y_train, y_test = train_test_split(
    X_class,
    y_class,
    test_size=0.2,
    random_state=42
)

log_model = LogisticRegression()

log_model.fit(X_train, y_train)

print("Coefficient:", log_model.coef_)
print("Intercept:", log_model.intercept_)
```

```python
probabilities = log_model.predict_proba(X_test)

print(probabilities)
```

```python
y_pred = log_model.predict(X_test)

print("Predicted classes:")
print(y_pred)

print("\nActual classes:")
print(y_test.to_numpy())
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

accuracy = accuracy_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)

print("Accuracy:", accuracy)
print("\nConfusion Matrix:")
print(cm)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Precision:", precision)
print("Recall:", recall)
print("F1-score:", f1)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

models = {
    "Logistic Regression": LogisticRegression(),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "SVM": SVC(kernel="linear"),
    "Naive Bayes": GaussianNB()
}

for name, model in models.items():
    model.fit(X_train, y_train)
    accuracy = model.score(X_test, y_test)

    print(f"{name}: {accuracy:.2f}")
```

```python
from sklearn.model_selection import cross_val_score

for name, model in models.items():

    scores = cross_val_score(
        model,
        X_class,
        y_class,
        cv=5,
        scoring="accuracy"
    )

    print(
        f"{name}: "
        f"Mean CV Accuracy = {scores.mean():.3f}"
    )
```
