```python
import matplotlib.pyplot as plt
```

```python
students = ["Anjula", "Anjali", "Tharindu", "Savindi"]
marks = [70, 80, 90, 100]

plt.bar(students,marks)
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```
