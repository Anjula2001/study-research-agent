```python
import matplotlib.pyplot as plt
```

```python
X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y)
plt.show()
```

```python
students = ["Anjula", "Anjali", "Tharindu", "Savindi"]
marks = [70, 80, 90, 100]

plt.bar(students,marks)
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python

hours = [1, 2, 3, 4, 5, 6]
marks = [45, 50, 60, 65, 75, 85]

plt.scatter(hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
marks = [45, 50, 52, 55, 60, 62, 65, 70, 72, 75, 80, 85, 90, 92, 95]

plt.hist(marks)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks")

plt.show()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temperature = [28, 30, 29, 31, 30]

plt.plot(days, temperature)

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature")

plt.grid()

plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]

anjula_marks = [60, 65, 70, 75, 80]
anjali_marks = [70, 72, 75, 78, 85]

plt.plot(months, anjula_marks, label="Anjula")
plt.plot(months, anjali_marks, label="Anjali")

plt.xlabel("Month")
plt.ylabel("Marks")
plt.title("Student Marks Progress")

plt.legend()
plt.grid()

plt.show()
```

```python
import pandas as pd

data = {
    "Name": ["Anjula", "Anjali", "Tharindu", "Savindi"],
    "Age": [25, 25, 29, 31],
    "Marks": [70, 80, 90, 100]
}

df = pd.DataFrame(data)

plt.bar(df["Name"], df["Marks"])

plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
```
