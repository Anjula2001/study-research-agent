```python
import matplotlib.pyplot as plt
```

```python
X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y)
plt.show()
```

```python
students = ["Anjula", "Anjali", "Tharindu", "Savindi"]
marks = [70, 80, 90, 100]

plt.bar(students,marks)
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python

hours = [1, 2, 3, 4, 5, 6]
marks = [45, 50, 60, 65, 75, 85]

plt.scatter(hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
marks = [45, 50, 52, 55, 60, 62, 65, 70, 72, 75, 80, 85, 90, 92, 95]

plt.hist(marks)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks")

plt.show()
```
